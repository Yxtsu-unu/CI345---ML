```python
# 17) + 18) Show the number of employees for each performance level 
# and create pie chart (and show percentage)



avg_salary_dept = df.groupby('Department')['Salary'].mean()
avg_salary_dept.plot(kind='bar', title='Average Salary by Department')
plt.ylabel('Average Salary')
plt.xticks(rotation=45)
plt.show()
```

```python
# 19) + 20) 
# Show the number of employee for each department,
# and sorted by the values from hight to low
# And plot the result as bar chart

#  .groupby('Department') or df.Department.value_counts() 
#  .sort_values(ascending=False)  
#  .plot 


perf_counts = df['Performance'].value_counts()
print(perf_counts)

perf_counts.plot(kind='pie', autopct='%1.1f%%', title='Employees by Performance')
plt.ylabel('')
plt.show()

```

```python
dept_counts = df['Department'].value_counts().sort_values(ascending=False)
print(dept_counts)

dept_counts.plot(kind='bar', title='Employees by Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.show()
```
